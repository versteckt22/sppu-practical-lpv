```python
# ==========================================
# Google Stock Price Prediction using RNN
# From Scratch
# ==========================================

# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# ==========================================
# LOAD DATASET
# ==========================================

# Load CSV File
dataset = pd.read_csv("Google_Stock_Price_Train.csv")

# Display First 5 Rows
print(dataset.head())

# ==========================================
# SELECT FEATURE AND TARGET
# ==========================================

# Using Open price column
data = dataset[['Open']].values

# ==========================================
# NORMALIZATION
# ==========================================

# Convert values between 0 and 1
scaler = MinMaxScaler(feature_range=(0,1))

data_scaled = scaler.fit_transform(data)

# ==========================================
# CREATE SEQUENCES
# ==========================================

x = []
y = []

# Use previous 60 days to predict next day
for i in range(60, len(data_scaled)):

    x.append(data_scaled[i-60:i, 0])
    y.append(data_scaled[i, 0])

# Convert to numpy arrays
x = np.array(x)
y = np.array(y)

# ==========================================
# RESHAPE FOR RNN
# ==========================================

x = np.reshape(x, (x.shape[0], x.shape[1], 1))

# ==========================================
# TRAIN TEST SPLIT
# ==========================================

x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

# ==========================================
# BUILD RNN MODEL
# ==========================================

model = Sequential()

# RNN Layer
model.add(
    SimpleRNN(
        units=50,
        activation='tanh',
        input_shape=(x_train.shape[1], 1)
    )
)

# Output Layer
model.add(Dense(1))

# ==========================================
# COMPILE MODEL
# ==========================================

model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)

# ==========================================
# TRAIN MODEL
# ==========================================

model.fit(
    x_train,
    y_train,
    epochs=20,
    batch_size=32
)

# ==========================================
# PREDICTION
# ==========================================

predicted = model.predict(x_test)

# Convert back to original values
predicted = scaler.inverse_transform(predicted)

# Convert actual values
actual = scaler.inverse_transform(y_test.reshape(-1,1))

# ==========================================
# DISPLAY RESULTS
# ==========================================

print("\nPredicted vs Actual Prices:\n")

for i in range(10):

    print("Actual Price =", actual[i][0])
    print("Predicted Price =", predicted[i][0])
    print()

# ==========================================
# PLOT GRAPH
# ==========================================

plt.plot(actual, color='red', label='Actual Price')

plt.plot(predicted, color='blue', label='Predicted Price')

plt.title("Google Stock Price Prediction")

plt.xlabel("Time")
plt.ylabel("Stock Price")

plt.legend()

plt.show()
```
